

1.   The current approach is insufficient b/c the random sample doesn't have memory of past selection. Meaning some people could never be picked, and some might be picked each class. Not a very fair scenerio.

2.   The fairest outcome would be for each student to go once, but that's not possible due to the # of students. The most fair and possible outcome would be for 5 students to be selected twice, and 6 students to be selected once.

3.   Proposed statregy:
   - Track # times each student is picked (max of 2)
   - Prioritize students who haven't been picked yet
   - Fill other spots by students only picked once
   - Randomly sample from pool to prevent duplicates in session
   - After 4 sessions, 16 spots are filled. So all 11 students go at least once, and some go twice.


```python
import random
from collections import Counter

# student list of names
students = ['Alice','Bob','Charlie','Diana','Ethan','Fiona','George','Hannah','Ian','Jenny','Kevin']

# track # times each is picked
selection_count = {name: 0 for name in students}

sessions = []

for session_num in range(4):
   # prioritize students with zero selections 1st
    never_selected = [s for s in students if selection_count[s] == 0]
    selected_once = [s for s in students if selection_count[s] == 1]

    session = []

    # fill the pool w/ never selected students 1st
    if never_selected:
        need = min(4, len(never_selected))
        session.extend(random.sample(never_selected, need))

    # fill the rest of the spots w/ students selected once
    remaining_spots = 4 - len(session)
    if remaining_spots > 0 and selected_once:
        session.extend(random.sample(selected_once, min(remaining_spots, len(selected_once))))

    # update the counts
    for name in session:
        selection_count[name] += 1

    sessions.append(session)
    print(f"Session {session_num + 1}: {session}")

print("\nSelection cCunts:")
for name in students: print(f"{name}: {selection_count[name]} time(s)")
```
