```python
import pandas as pd
from google.colab import drive
from collections import Counter
import re

# load file
df = pd.read_csv('/content/Assignment2_Part2_Dataset.csv')

# preprocess text
def clean_text(text):
    text = str(text).lower()    # convert to lowercase
    text = re.sub(r'[^a-z\s]','',text)    # get rid of special characthers
    return [w for w in text.split() if len(w) > 2]    # get rid of short words

# get top 10 words for each class
def analyze_class(df, class_num):
    tweets = df[df['class'] == class_num]['tweet']  # filter down to 1 class
    all_words = []
    for tweet in tweets: all_words.extend(clean_text(tweet))  #clean/add words to list

    word_counts = Counter(all_words)  # counts freq. of each word
    total = sum(word_counts.values()) # counts total words

    results = []
    # get top 10
    for word, count in word_counts.most_common(10):results.append({'Word': word,'Frequency': count,'Probability': count/total})

    return pd.DataFrame(results), total

# analyze classes
classes = {0: 'Hate Speech', 1: 'Offensive Language', 2: 'Neither'}

# Hate Speech
results_hate, total_hate = analyze_class(df, 0)
print("Hate Speech (Total Words: {:,})".format(total_hate))
print(results_hate.to_string(index=False))

# Offensive Language
results_off, total_off = analyze_class(df, 1)
print("\nOffensive Language (Total Words: {:,})".format(total_off))
print(results_off.to_string(index=False))

# Neither
results_neither, total_neither = analyze_class(df, 2)
print("\nNeither (Total Words: {:,})".format(total_neither))
print(results_neither.to_string(index=False))
```

**Observations:**
- There was some overlap between all groups due to commonly used words (that, like, and, the, you)
- Hate speech had a lot of curse words in top 10, mostly slurs
- Offensive language were semi-curse words. Offensive but not hate speech. This was largest category
- Neither category had a lot of generic words that didn't include curse words/slurs at all
