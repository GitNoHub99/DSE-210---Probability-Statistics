```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt


# Define delivery time elements (X1-X5) as normal distributions
x1 = norm(1, 0.2)
x2 = norm(5, 0.4)
x3 = norm(2, 0.02)
x4 = norm(2, 0.7)
x5 = norm(0.5, 0.6)

# X stats
x_mean = x1.mean() + x2.mean() + x3.mean() + x4.mean() + x5.mean()
x_var = x1.var() + x2.var() + x3.var() + x4.var() + x5.var()
x_std = np.sqrt(x_var)

print(f"X~Normal({x_mean}, {x_var:.2f})\n")

# Question #1: Plot
samples = norm.rvs(x_mean, x_std, size=10000)
plt.hist(samples, bins=50, density=True, alpha=0.7, edgecolor='black')
plt.xlabel('Total Delivery Time (Hrs)')
plt.ylabel('Density')
plt.title('Distribution of Delivery Time from A to B')
plt.show()

# Question #2: Expected revenue - E[Revenue] = SumOf(outcome) * P(outcome)
prob_within_13 = norm.cdf(13, x_mean, x_std)
expected_revenue = prob_within_13*1000 + (1-prob_within_13)*-50

print(f"\nProbability of Delivery Time within 13 hrs: {prob_within_13:.2f}")
print(f"Expected Revenue: ${expected_revenue:.2f}")
```
